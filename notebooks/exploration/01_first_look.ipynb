```python
import pandas as pd

df = pd.read_csv('../../data/raw/hotel_bookings.csv')
df.shape
```

```python
import sqlite3

conn = sqlite3.connect('../../data/processed/stayiq.db')
df.to_sql('bookings', conn, if_exists='replace', index=False)
print("Table created successfully")
```

```python
%load_ext sql
%sql sqlite:///../../data/processed/stayiq.db
```

```python
%%sql
SELECT lead_time
FROM bookings
WHERE is_canceled = 1
LIMIT 10;
```

```python
query = """
SELECT lead_time
FROM bookings
WHERE is_canceled = 1
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT hotel, adr
FROM bookings
WHERE is_canceled = 0 AND adr > 100
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)
result
```

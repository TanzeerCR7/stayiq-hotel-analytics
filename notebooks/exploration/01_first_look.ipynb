```python
import pandas as pd

df = pd.read_csv('../../data/raw/hotel_bookings.csv')
df.shape
```

```python
import sqlite3

conn = sqlite3.connect('../../data/processed/stayiq.db')
df.to_sql('bookings', conn, if_exists='replace', index=False)
print("Table created successfully")
```
